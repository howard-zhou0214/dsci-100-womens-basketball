#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
# Use seed and reg_wins_pct as numeric predictors of tournament success.
# In the supplied sample, conf_wins_pct is mostly missing, whereas
# reg_wins_pct is available; choosing it avoids those conference-data gaps.
# Tournament results are used only to define the response, not as predictors.

# The source is already tidy: each row is one team's tournament appearance,
# and each column represents a variable, so no pivot is needed.
basketball_tidy <- basketball_raw %>%
  # Keep the two predictors and the count needed to create the response.
  select(seed, reg_wins_pct, tourney_wins) %>%
  # Exclude appearances missing any selected value; do not replace missing
  # outcomes with zero. Conclusions will apply to complete selected records.
  drop_na(seed, reg_wins_pct, tourney_wins) %>%
  # Seed 0 denotes special opening-round teams in 1983, not a standard rank.
  # Restrict the analysis to positive seeds instead of ranking 0 above 1.
  filter(seed > 0) %>%
  # Create a categorical response: yes means at least one tournament win.
  mutate(
    won_tournament_game = factor(
      if_else(tourney_wins >= 1, "yes", "no"),
      levels = c("no", "yes")
    )
  ) %>%
  # Drop tourney_wins to prevent the outcome count from becoming a predictor.
  # Keep the win percentage in its original percentage-point units.
  select(seed, reg_wins_pct, won_tournament_game)

# Show the first six rows with two numeric variables and one categorical one.
head(basketball_tidy, 6)
```

Each point shows the share of teams at a given seed that won at least one tournament game. Comparing shares accounts for differences in how often each seed appeared. A change in the win rate across seeds would suggest that seed helps distinguish the two outcomes.

```r
# Within each seed, the mean of a TRUE/FALSE indicator is the share of wins.
basketball_tidy %>%
  group_by(seed) %>%
  summarise(win_share = mean(won_tournament_game == "yes"), .groups = "drop") %>%
  ggplot(aes(x = seed, y = win_share)) +
  geom_point(size = 2.5, colour = "steelblue4") +
  scale_x_continuous(breaks = sort(unique(basketball_tidy$seed))) +
  scale_y_continuous(labels = scales::label_percent(), limits = c(0, 1)) +
  labs(title = "Tournament success by seed",
       x = "Tournament seed (1 is strongest)",
       y = "Share winning at least one game") +
  theme_minimal()
```
